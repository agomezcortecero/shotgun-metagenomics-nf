```python
import pandas as pd
import matplotlib.pyplot as plt

# Path to your Excel file
file_path = 'all_bacteria2.xlsx'

# Load the Excel file
xls = pd.ExcelFile(file_path)

# Get the sheet names
sheet_names = xls.sheet_names

# Define a more professional-looking color palette for each sample group
colors = {'Treatment1': '#4C72B0', 'Untreated': '#DD8452', 'Treatment2': '#55A868'}

fig, ax = plt.subplots(figsize=(18, 10))

# Initialize offset and bar width for plotting
location_offset = 0
bar_width = 0.35
xtick_positions = []  # To store the positions for x-ticks
xtick_labels = []  # To store the labels for x-ticks
tab_labels = []  # To hold labels for the tabs
tab_positions = []  # To hold positions for tab labels
legend_labels = []  # To keep track of the labels already added to the legend

# Iterate over each sheet to process and plot data
for sheet_name in sheet_names:
    # Load the data from the current sheet
    data = pd.read_excel(file_path, sheet_name=sheet_name)

    # Determine which label to use based on the sheet name
    if "Untreated" in sheet_name:
        control_label = "Untreated"
    else:
        control_label = "Treatment2"

    # Identify columns for "Treatment1" and the control group (either "Untreated" or "Treatment2")
    treatment1_columns = [col for col in data.columns if 'Treatment1' in col]
    control_columns = [col for col in data.columns if control_label in col]

    # Calculate average scores for Treatment1 and the control group
    data['Average_Treatment1'] = data[treatment1_columns].mean(axis=1)
    data[f'Average_{control_label}'] = data[control_columns].mean(axis=1)

    # Calculate score differences and sort by these differences
    data['Score_Difference'] = abs(data['Average_Treatment1'] - data[f'Average_{control_label}'])
    data_sorted = data.sort_values(by='Score_Difference', ascending=False)

    # Set x positions for the groups of bars after sorting
    x_positions = range(location_offset, location_offset + len(data_sorted['Species']))
    xtick_positions.extend(x_positions)  # Assign each species' position
    xtick_labels.extend(data_sorted['Species'])  # Use only species names for labels

    # Record positions for tab labels
    tab_labels.append(sheet_name)
    tab_positions.append(sum(x_positions) / len(x_positions))  # Calculate midpoint for tab label

    # Plot bars for average Treatment1
    if 'Treatment1' not in legend_labels:
        ax.bar(x_positions, data_sorted['Average_Treatment1'], width=bar_width, label='Treatment1', color=colors['Treatment1'])
        legend_labels.append('Treatment1')
    else:
        ax.bar(x_positions, data_sorted['Average_Treatment1'], width=bar_width, color=colors['Treatment1'])

    # Plot bars for average Untreated or Treatment2
    if control_label not in legend_labels:
        ax.bar([p + bar_width for p in x_positions], data_sorted[f'Average_{control_label}'], width=bar_width, label=control_label, color=colors[control_label])
        legend_labels.append(control_label)
    else:
        ax.bar([p + bar_width for p in x_positions], data_sorted[f'Average_{control_label}'], width=bar_width, color=colors[control_label])

    # Update the location offset for the next sheet's data
    location_offset += len(data_sorted['Species']) + 1  # Adding a space between groups from different sheets

# Setting the x-ticks to be in the middle of the groups of bars
ax.set_xticks([p + bar_width / 2 for p in xtick_positions])
ax.set_xticklabels(xtick_labels, rotation=45, ha='right')

# Italicize and bold labels for all species, with special treatment for "Bacillus"
for ticklabel in ax.get_xticklabels():
    if "Bacillus" in ticklabel.get_text():
        ticklabel.set_fontweight('bold')
    ticklabel.set_fontstyle('italic')

# Adding sheet names above the plot with 45-degree rotation
# Adding sheet names above the plot with slight adjustment to the right
for label, pos in zip(tab_labels, tab_positions):
    ax.text(
        pos + 2,  # Add a small offset to move the label to the right
        ax.get_ylim()[1] * 1.05,
        label,
        ha='center',  # Keep it centered after applying the offset
        va='bottom',
        rotation=45
    )


# Adding labels and legend
ax.set_xlabel('Species')
ax.set_ylabel('Average Score')
ax.legend(title="Treatment Groups")

# Save the plot as a high-quality image (png format, 300 dpi)
plt.tight_layout()
plt.savefig('species_abundance_plot.png', dpi=300)

# Show the plot
plt.show()
```
